```python

# Tanglish Code-Switching Analyzer

import re
from collections import Counter

# Sample vocabulary for a basic prototype
tamil_words = {
    "enna", "epdi", "iruku", "irukku", "panna",
    "venum", "macha", "seri", "romba", "nalla",
    "kudu", "kudunga", "pogalam", "theriyum"
}

english_words = {
    "project", "code", "python", "meeting", "today",
    "report", "complete", "class", "task", "submit",
    "working", "finish", "please", "tomorrow"
}

def analyze_tanglish(text):
    words = re.findall(r"\b[a-zA-Z]+\b", text.lower())

    tamil_found = [w for w in words if w in tamil_words]
    english_found = [w for w in words if w in english_words]
    unknown = [
        w for w in words
        if w not in tamil_words and w not in english_words
    ]

    total = len(words)
    tamil_percentage = len(tamil_found) / total * 100 if total else 0
    english_percentage = len(english_found) / total * 100 if total else 0

    # Count language switches between recognized Tamil and English words
    recognized = []
    for word in words:
        if word in tamil_words:
            recognized.append("Tamil")
        elif word in english_words:
            recognized.append("English")

    switches = sum(
        1 for i in range(1, len(recognized))
        if recognized[i] != recognized[i - 1]
    )

    print("=" * 45)
    print("TANGLISH CODE-SWITCHING ANALYZER")
    print("=" * 45)
    print("Input:", text)
    print("\nTotal words:", total)
    print("Recognized Tamil words:", len(tamil_found))
    print("Recognized English words:", len(english_found))
    print(f"Tamil vocabulary share: {tamil_percentage:.1f}%")
    print(f"English vocabulary share: {english_percentage:.1f}%")
    print("Language switches:", switches)

    print("\nFrequent words:")
    for word, count in Counter(words).most_common(5):
        print(f"{word}: {count}")

    print("\nUnclassified words:", ", ".join(unknown) or "None")
    print("=" * 45)


# Sample input
text = (
    "Macha project work nalla iruku. "
    "Python code complete panna venum today."
)

analyze_tanglish(text)
```
